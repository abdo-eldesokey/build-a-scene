```python
from pathlib import Path
import json
import numpy as np

metric_json_path = Path("../workspace/build-a-scene/generation/metrics")

clip_avg = []
clip_max = []
moa_avg = []
moa_max = []
miou = []
for json_fname in list(metric_json_path.glob("*.json")):
    file = open(json_fname, "r")
    metrics = json.load(file)
    # if metrics["yolo_metrics"]["per_seed"]
    clip_avg.append(metrics["clip"]["clip_avg"])
    clip_max.append(metrics["clip"]["clip_max"])
    moa_avg.append(metrics["yolo_metrics"]["mean_oa"])
    moa_max.append(metrics["yolo_metrics"]["max_oa"])
    miou.append(metrics["yolo_metrics"]["mean_iou"])

print("Average CLIP: ", np.nanmean(np.array(clip_avg)))
print("Max CLIP: ", np.nanmean(np.array(clip_max)))
print("Average mOA: ", np.nanmean(np.array(moa_avg)))
print("Max mOA: ", np.nanmean(np.array(moa_max)))
print("Average mIOU: ", np.nanmean(np.array(miou)))
```

```python
from pathlib import Path
import json
import numpy as np

metric_json_path = Path("../workspace/build-a-scene/consistency/metrics")

clip_max = []
clip_avg = []
ssim_max = []
ssim_avg = []
psnr_max = []
psnr_avg = []
which = "fg"
for json_fname in list(metric_json_path.glob("*.json")):
    file = open(json_fname, "r")
    metrics = json.load(file)
    clip_avg.append(metrics[which]["mean_clip"])
    clip_max.append(metrics[which]["max_clip"])
    ssim_avg.append(metrics[which]["mean_ssim"])
    ssim_max.append(metrics[which]["max_ssim"])
    psnr_avg.append(metrics[which]["mean_psnr"])
    psnr_max.append(metrics[which]["max_psnr"])

print("Average CLIP: ", np.nanmean(np.array(clip_avg)))
print("Max CLIP: ", np.nanmean(np.array(clip_max)))
print("Average SSIM: ", np.nanmean(np.array(ssim_avg)))
print("Max SSIM: ", np.nanmean(np.array(ssim_max)))
print("Average PSNR: ", np.nanmean(np.array(psnr_avg)))
print("Max PSNR: ", np.nanmean(np.array(psnr_max)))
```
